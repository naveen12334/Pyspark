Reading CSV files from Volume

```python
#read csv files from volume
dfcsv = spark.read.csv("/Volumes/my_files/my_schema/csv_files/my_files/sales.csv", header=True)
display(dfcsv)
```

```python
#Make header = True
dfcsv = spark.read.option("header","true").csv("/Volumes/my_files/my_schema/csv_files/my_files/sales.csv")
                          
display(dfcsv)
```

```python
#use inferSchema to change columns data type automatically according to schema
dfcsv = spark.read.\
    option("header","true")\
        .option("inferSchema","true")\
    .csv("/Volumes/my_files/my_schema/csv_files/my_files/sales.csv")
                          
display(dfcsv)
```

```python
#schema ourself

schema_str = "id INT, customer_name STRING, product STRING, quantity INT, price INT, remarks STRING"
dfcsv = spark.read.\
    schema(schema_str)\
    .csv("/Volumes/my_files/my_schema/csv_files/my_files/sales.csv")
                          
display(dfcsv)
dfcsv.printSchema()
dfcsv = spark.read
```

```python
#modular own schema using python objects

from pyspark.sql.types import StructType, StructField, StringType, IntegerType,DoubleType

schema = StructType([
    StructField("id", IntegerType(), True),
    StructField("customer_name", StringType(), True),
    StructField("product", StringType(), True),
    StructField("quantity", IntegerType(), True),
    StructField("price", DoubleType(), True),
    StructField("remarks", StringType(), True)
])
dfcsv = spark.read\
    .schema(schema)\
    .csv("/Volumes/my_files/my_schema/csv_files/my_files/sales.csv")
                          
display(dfcsv)
dfcsv.printSchema()
```

```python
#reading csv files with delimiter distribution

df = spark.read.\
    option("header","true")\
        .option("inferSchema","true")\
            .option("delimiter","|")\
            .csv("/Volumes/my_files/my_schema/csv_files/my_files/sales_pipe.csv")
display(df)
```

```python
#handling single quotes

df = spark.read.\
    option("header","true")\
        .option("inferSchema","true")\
            .option("delimiter","|")\
                .option("quote","'")\
            .csv("/Volumes/my_files/my_schema/csv_files/my_files/sales_pipe_single_quote.csv")
display(df)
```

```python
#handling single quotes with escape

df = spark.read.\
    option("header","true")\
        .option("inferSchema","true")\
            .option("delimiter","|")\
                .option("quote","'")\
                    .option("escape","#")\
            .csv("/Volumes/my_files/my_schema/csv_files/my_files/sales_pipe_single_quote.csv")
display(df)
```

```python
#handling single quotes in multiple lines

df = spark.read.\
    option("header","true")\
        .option("inferSchema","true")\
            .option("delimiter","|")\
                .option("quote","'")\
                    .option("escape","#")\
                        .option("multiLine","true")\
            .csv("/Volumes/my_files/my_schema/csv_files/my_files/sales_pipe_single_quote.csv")
display(df)
```

**Spark - Handle Corrupted Data**

Option 1 - Read Mode

1. PERMISSIVE - Default
2. FAILFAST - Throw error and stops reading
3. DROPMALFORMED - Silenty drops rows

Option 2 - Path to write corrupted records directly

badRecordsPath - Path to write corrupted records directly

```python
#default json read file

df = spark.read.json("/Volumes/my_files/my_schema/json_files/sales_corrupted.json")
display(df)
```

```python
#dropped corrupted records

df = spark.read.format("json").\
    option("mode","DROPMALFORMED").\
        load("/Volumes/my_files/my_schema/json_files/sales_corrupted.json")
display(df)
```

```python
#failfast mode

df = spark.read.format("json").\
    option("mode","FAILFAST").\
        load("/Volumes/my_files/my_schema/json_files/sales_corrupted.json")
display(df)
```

```python
#Permissive mode

df = spark.read.format("json").\
    option("mode","PERMISSIVE").\
        load("/Volumes/my_files/my_schema/json_files/sales_corrupted.json")
display(df)
```

Handle corrupted csv files

```python
#read csv file

df = spark.read.format("csv").load("/Volumes/my_files/my_schema/csv_files/my_files/sales_corrupted.csv")

display(df)
```

```python
#define schema
df = spark.read.\
    schema("id integer, customer_name string, product string, quantity integer, price double, remarks string,corrupt_record string")\
        .format("csv")\
            .option("mode","PERMISSIVE")\
        .load("/Volumes/my_files/my_schema/csv_files/my_files/sales_corrupted.csv")

display(df)
```

```python
#badrecords path
df = spark.read.\
    schema("id integer, customer_name string, product string, quantity integer, price double, remarks string,corrupt_record string")\
        .format("csv")\
            .option("badRecordsPath","/Volumes/my_files/my_schema/badRecordsPAth")\
        .load("/Volumes/my_files/my_schema/csv_files/my_files/sales_corrupted.csv")

display(df)
```

```python
#recursiveFileLookup (by default it will check badrecords path)
df2 = spark.read.format("json").option("recursiveFileLookup","true").load("/Volumes/my_files/my_schema/badRecordsPAth")
display(df2)
```
